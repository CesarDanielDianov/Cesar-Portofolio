```python
import numpy as np
np.set_printoptions(precision=6, suppress=True)

def mostra(titulo, comando):
    """Imprime o título, o comando usado e o resultado."""
    print(f"{titulo}")
    print(f">>> {comando}")
    resultado = eval(comando)
    print(resultado, "\n")
    return resultado

# Dados do enunciado
x = mostra("Inputs x:", "np.array([0, 1, -2, 5, -4])")
t = mostra("Targets t:", "np.array([-1, 1, -1, 0, -1])")

# (a) Matriz de design
Phi = mostra("(a) Matriz de design Φ  [colunas: 1, |x|, √|x|, ∛|x|]:",
             "np.column_stack([np.ones(5), np.abs(x), np.abs(x)**(1/2), np.abs(x)**(1/3)])")

# (b) w = (ΦᵀΦ)⁻¹ Φᵀ t
PhiT_Phi = mostra("ΦᵀΦ:", "Phi.T @ Phi")
PhiT_t   = mostra("Φᵀt:", "Phi.T @ t")
inv      = mostra("(ΦᵀΦ)⁻¹:", "np.linalg.inv(PhiT_Phi)")
w        = mostra("(b) Pesos w = (ΦᵀΦ)⁻¹ Φᵀt:", "inv @ PhiT_t")

# Verificação
y = mostra("Previsões y = Φw:", "Phi @ w")
E = mostra("Erro E(w) = ½ Σ(t − y)²:", "0.5 * np.sum((t - y)**2)")
```
